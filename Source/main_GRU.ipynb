load dataset functions

```python
def parse_slots(annot_utt):
    # get list of (slot_type, slot_span) from annot_utt
    # turns the [color_type : pink] into ("color_type", "pink")
    return [(t.strip(), s.strip()) for t, s in SLOT_REGEX.findall(annot_utt)]

def load_massive(path=DATA_PATH):
    splits = {"train": [], "dev": [], "test": []}
    with open(path, encoding="utf-8") as f:
        for line in f:
            if not line.strip():
                continue
            r = json.loads(line)
            splits[r["partition"]].append({
                "id": r["id"],
                "scenario": r["scenario"],
                "intent": r["intent"],
                "utt": r["utt"],
                "annot_utt": r["annot_utt"],
                "slots": parse_slots(r["annot_utt"]),  
            })
    return splits
```

preprocessing data

going with BIOES tagging for the parts of the sentence

```python
def annot_to_bioes(annot_utt):
    """Convert annotated text into tokens and BIOES tags."""
    # use whitespace to split, so just do utt.split()
    # then assign tags based on the slots
    # theres 55 slot types, so they will be tagged as B-time, I-time, B-house_place that kind
    # the 55 types are defined in the above cell

    tokens = []
    tags = []
    position = 0

    # Process each annotated slot, such as [time : five am]
    for match in SLOT_REGEX.finditer(annot_utt):

        # Words outside a slot are tagged as O
        outsideText = annot_utt[position:match.start()]

        for word in outsideText.split():
            tokens.append(word)
            tags.append("O")

        slotType = match.group(1).strip()
        words = match.group(2).split()

        # Assign BIOES tag for each word 
        for i, word in enumerate(words):
            if len(words) == 1:
                tag = f"S-{slotType}"
            elif i == 0:
                tag = f"B-{slotType}"
            elif i == len(words) - 1:
                tag = f"E-{slotType}"
            else:
                tag = f"I-{slotType}"

            tokens.append(word)
            tags.append(tag)

        position = match.end()

    # Process any remaining words after the last slot
    for word in annot_utt[position:].split():
        tokens.append(word)
        tags.append("O")

    return tokens, tags


def build_tag_vocab(records):

    slotTypes = sorted({
        slotType
        for record in records
        for slotType, _ in record["slots"]
    })

    tags = ["O", PAD_TOKEN]

    for slotType in slotTypes:
        for prefix in ("B", "I", "E", "S"):
            tags.append(f"{prefix}-{slotType}")

    tagToId = {tag: i for i, tag in enumerate(tags)}

    return tags, tagToId

def build_word_vocab(records):
    words = [PAD_TOKEN, UNK_TOKEN]

    newSet = set()
    for record in records:
        for inner in record["utt"].split():
            newSet.add(inner)
    newSet = sorted(newSet)

    for word in newSet:
        words.append(word)
    wordToId = {word: i for i, word in enumerate(words)}

    return words, wordToId

def build_intent_vocab(records):
    unique_records = set()
    for record in records:
        unique_records.add(record["intent"])

    intents = sorted(unique_records)

    intentToId = {label: i for i, label in enumerate(intents)}
    return intents, intentToId

```

Decode the BIOES tags into (slot_type, slot_span)

```python
def decode_bioes(tokens, tags):
    """Convert BIOES tags into a list of (slotType, slotSpan) pairs."""

    spans = []
    currentType = None
    currentTokens = []

    def closeSpan():
        nonlocal currentType, currentTokens

        if currentType is not None:
            slotText = " ".join(currentTokens)
            spans.append((currentType, slotText))

        currentType = None
        currentTokens = []

    for token, tag in zip(tokens, tags):

        if tag == "O":
            closeSpan()
            continue

        prefix, slotType = tag.split("-", 1)

        if prefix == "B":
            closeSpan()
            currentType = slotType
            currentTokens = [token]

        elif prefix == "I":
            if currentType == slotType:
                currentTokens.append(token)
            else:
                closeSpan()
                currentType = slotType
                currentTokens = [token]

        elif prefix == "E":
            if currentType == slotType:
                currentTokens.append(token)
                closeSpan()
            else:
                closeSpan()
                spans.append((slotType, token))

        elif prefix == "S":
            closeSpan()
            spans.append((slotType, token))

        else:
            closeSpan()

    # Include any unfinished span at the end
    closeSpan()

    return spans
```

Evaluation functions

```python
def eval_intent_accuracy(actual, predicted):
    """Calculate the percentage of correctly predicted intents."""

    assert len(actual) == len(predicted)

    correctCount = sum(
        gold == pred
        for gold, pred in zip(actual, predicted)
    )

    return 100.0 * correctCount / len(actual)

def eval_macro_f1(actual, predicted):
    """Calculate micro F1 for slot types and spans across all utterances."""

    assert len(actual) == len(predicted)

    truePositives = 0
    falsePositives = 0
    falseNegatives = 0

    for act, pred in zip(actual, predicted):
        actualCount = Counter(act)
        predCount= Counter(pred)

        matchedCount = sum((actualCount & predCount).values())

        truePositives += matchedCount
        falsePositives += sum(predCount.values()) - matchedCount
        falseNegatives += sum(actualCount.values()) - matchedCount

    denominator = (
        2 * truePositives
        + falsePositives
        + falseNegatives
    )

    if denominator == 0:
        return 0.0

    return 100.0 * 2 * truePositives / denominator

def evaluate(goldIntents, predIntents, goldSlots, predSlots):
    """Calculate and return intent accuracy and slot F1."""

    intentAcc = round(eval_intent_accuracy(goldIntents, predIntents), 2)

    slotF1 = round(eval_macro_f1(goldSlots, predSlots), 2)

    return {
        "intent_acc": intentAcc,
        "slot_f1": slotF1,
    }
```

Dataclass

```python
class MassiveDataset(Dataset):
    def __init__(self, records, wordToId, tagToId, intentToId):
        self.records = records
        self.wordToId = wordToId
        self.tagToId = tagToId
        self.intentToId = intentToId 

    def __len__(self):
        return len(self.records)
    
    def __getitem__(self, idx):
        record = self.records[idx]
        word_ids = []
        for r in record["utt"].split():
            if self.wordToId.get(r):
                word_ids.append(self.wordToId[r])
            else:
                word_ids.append(self.wordToId[UNK_TOKEN])

        tokens, tags = annot_to_bioes(record["annot_utt"])
        tag_ids = []

        for tag in tags:
            tag_ids.append(self.tagToId[tag])

        intent_id = self.intentToId[record["intent"]]

        return word_ids, tag_ids, intent_id

```

```python
def collate_fn(batch):
    word_id_lst = []
    tag_id_lst = []
    intent_id_lst = []

    longest = [0, 0, 0]
    for b in batch:
        w, t, i = b
        longest[0] = max(longest[0], len(w))
        longest[1] = max(longest[1], len(t))
        word_id_lst.append(w)
        tag_id_lst.append(t)
        intent_id_lst.append(i)

    for w in word_id_lst:
        w += (longest[0] - len(w)) * [wordToId[PAD_TOKEN]]
    for t in tag_id_lst:
        t += (longest[1] - len(t)) * [tagToId[PAD_TOKEN]]

    return torch.tensor(word_id_lst), torch.tensor(tag_id_lst), torch.tensor(intent_id_lst)
```

GRU 

```python
import torch.nn as nn

class GRUModel(nn.Module):
    def __init__(self, vocab_size, num_tags, num_intents):
        super().__init__()
        self.embedding = nn.Embedding(vocab_size, EMBEDDING_DIM, padding_idx=wordToId[PAD_TOKEN])
        self.gru = nn.GRU(EMBEDDING_DIM, HIDDEN_SIZE, batch_first=True, bidirectional=True)
        self.intent_head = nn.Linear(2 * HIDDEN_SIZE, num_intents)
        self.slot_head = nn.Linear(2 * HIDDEN_SIZE, num_tags)

    def forward(self, input_ids):
        embedded = self.embedding(input_ids) 
        gru_out, hidden = self.gru(embedded)        

        concatenated = torch.cat([hidden[0], hidden[1]], dim=1)
        
        slot_logits = self.slot_head(gru_out)
        intent_logits = self.intent_head(concatenated)

        return intent_logits, slot_logits
```

```python
def evaluate_split(records):
    dev_ds = MassiveDataset(records, wordToId, tagToId, intentToId)
    dev_loader = DataLoader(dev_ds, batch_size=BATCH_SIZE, shuffle=False, collate_fn=collate_fn)

    model.eval()
    pred_intents, gold_intents = [], []
    pred_slots, gold_slots = [], []

    with torch.no_grad():
        for record in records:
            word_ids = []
            for w in record["utt"].split():
                if wordToId.get(w):
                    word_ids.append(wordToId[w])
                else:
                    word_ids.append(wordToId[UNK_TOKEN])

            input_tensor = torch.tensor([word_ids])
            intent_logits, slot_logits = model(input_tensor)

            predicted_id = intent_logits.argmax(dim=1).item()
            pred_intents.append(intents[predicted_id])
            gold_intents.append(record["intent"])

            predicted_tag_ids = slot_logits.argmax(dim=2)[0]
            predicted_tags = ["O" if tags[i.item()] == PAD_TOKEN else tags[i.item()] for i in predicted_tag_ids]

            tokens = record["utt"].split()
            predicted_spans = decode_bioes(tokens, predicted_tags)

            pred_slots.append(predicted_spans)
            gold_slots.append(record["slots"])

        total = 0
        for batch in dev_loader:
            intent_logits, slot_logits = model(batch[0])
            intent_loss = intent_criterion(intent_logits, batch[2])
            slot_loss = slot_criterion(slot_logits.reshape(-1, len(tags)), batch[1].reshape(-1))
            total_loss = intent_loss + slot_loss

            total += total_loss.item()

        average_loss = total / len(dev_loader) 


    return eval_intent_accuracy(gold_intents, pred_intents), eval_macro_f1(gold_slots, pred_slots), average_loss
```

```python
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
import time

torch.manual_seed(42)
model = GRUModel(len(words), len(tags), len(intents))
optimizer = torch.optim.Adam(model.parameters(), lr=LEARNING_RATE)

start = time.time()

training_loss = []
elapsed_time = []
dev_intents = []
dev_slot_f1s = []
dev_losses = []

best_score = -1
best_epoch = -1
for epoch in range(NUM_EPOCHS):
    running_total_loss = 0
    model.train()                          
    for batch in train_loader:
        optimizer.zero_grad()              
        intent_logits, slot_logits = model(batch[0])
        intent_loss = intent_criterion(intent_logits, batch[2])
        slot_loss = slot_criterion(slot_logits.reshape(-1, len(tags)), batch[1].reshape(-1))
        loss = intent_loss + slot_loss
        running_total_loss += loss.item()

        loss.backward()                   
        optimizer.step()
    end = time.time()
    elapsed = end - start 
    dev_intent, dev_slot_f1, dev_loss = evaluate_split(splits["dev"])
    training_loss.append(running_total_loss / len(train_loader))
    elapsed_time.append(elapsed)
    dev_intents.append(dev_intent)
    dev_slot_f1s.append(dev_slot_f1)
    dev_losses.append(dev_loss)

    print(f"Epoch {epoch}: loss={(running_total_loss / len(train_loader)):.4f}, time={elapsed:.1f}s, dev_intent={dev_intent:.4f}, dev_slot_f1={dev_slot_f1:.4f}, dev_loss={dev_loss:.4f}")
    score = (dev_intent + dev_slot_f1) / 2
    if score > best_score:
        best_score = score
        best_epoch = epoch
        torch.save(model.state_dict(), "../Results/gru_baseline_best.pt")
    if epoch - best_epoch >= 10:
        break

print(f"Best epoch: {best_epoch}, best score: {best_score}, checkpoint: ../Results/gru_baseline_best.pt")
```
